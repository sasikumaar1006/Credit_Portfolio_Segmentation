```python
# Dropping irrelevant/redundant columns
df = df.drop(columns=['experience_c', 'title', 'desc'])

print("Remaining columns:", df.columns.tolist())
print("New Dataset Shape:", df.shape)
```

```python
# Dropping the single row where zip_code is missing
df = df.dropna(subset=['zip_code'])

print(df.isnull().sum())
df.shape
```

```python
# Converting the emp_length column to numerical
import re
import numpy as np

def clean_emp_length(val):
    if pd.isna(val):
        return np.nan
    val_str = str(val).strip().lower()
    
    # Handle '< 1 year' as 0
    if '< 1' in val_str:
        return 0
    
    # Extract the numeric digits from the string (e.g., '10+ years' -> 10, '3 years' -> 3)
    match = re.search(r'\d+', val_str)
    if match:
        return int(match.group())
    
    return np.nan

# Apply the function to create a new clean numerical column
df['emp_length_n'] = df['emp_length'].apply(clean_emp_length)

# Verify the unique values and check the conversion
print(df[['emp_length', 'emp_length_n']].drop_duplicates().head(10))
```

```python
high_dti_count = (df['dti_n'] > 100).sum()
print(f"Number of rows with dti_n > 100: {high_dti_count}")

# Inspecting the percentage of these extreme rows
total_rows = len(df)
print(f"Percentage of extreme DTI rows: {(high_dti_count / total_rows) * 100:.4f}%")
```

```python
#Removing those extreme DTI rows as it is a very small percentage
df = df[df['dti_n'] <= 100]

print("DTI outliers handled successfully!")
```

```python
# Convert issue_d to datetime format
df['issue_d'] = pd.to_datetime(df['issue_d'], format='%b-%Y')

# Verify the data type and check the converted values
print(df['issue_d'].dtypes)
print(df['issue_d'].head())
```
